## From computational evidence to functional evidence

This workflow helps researchers choose what to study next rather than producing
one final score. GTEx helps select a tissue, HuBMAP helps select a cell model,
Pharos shows what is known about a protein and which research tools are
available, and ProtVar provides predictions that can inform protein-focused
experiments.

For the *TNNT2* VUS, the next step is to compare wild-type and D259A troponin
complexes across calcium concentrations. This experiment would test whether
p.Asp259Ala changes calcium-regulated thin-filament activity. Clinical
interpretation would also require checking whether the phenotype matches,
whether the variant segregates with disease, how common it is in populations,
and whether functional studies support an effect.


## A reusable prioritization checklist

- Name the exact variant and the source of its annotation.
- Ask what biological level each dataset measures.
- Keep missing data separate from a measured zero.
- Save dataset versions, identifiers, and retrieval dates for reproducibility.
- Keep druggability separate from disease relevance.
- End with a clear next step that can be tested.


## Suggested next steps and reading

These papers offer several ways to continue the analysis:

- **Variant prioritization:** [Stenton et al.
  (2024)](https://doi.org/10.1186/s40246-024-00604-w),
  co-authored by Elizabeth A. Worthey, compares rare-disease variant
  prioritization methods and the evidence used to rank candidate genetic
  variants.
- **Protein structure:** [Varga et al.
  (2025)](https://doi.org/10.1016/j.ajhg.2025.03.010) explains
  how protein structures can inform variant interpretation and how the choice
  of structural model affects the analysis.
- **Splicing and RNA evidence:** [Truty et al.
  (2021)](https://doi.org/10.1016/j.ajhg.2021.03.006) examines
  how RNA analysis can help interpret variants predicted to affect splicing.
- **Clinical WGS workflow:** [Worthey et al.
  (2012)](https://doi.org/10.1186/1753-6561-6-S6-O11)
  describes an early clinical whole-genome sequencing workflow that considered
  variant quality, functional effect, disease knowledge, and phenotype
  relevance together.
